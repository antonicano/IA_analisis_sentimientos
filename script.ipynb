## Análisis de sentimientos

```python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification

# Cargar el tokenizer y el modelo
tokenizer = AutoTokenizer.from_pretrained("edumunozsala/roberta_bne_sentiment_analysis_es")
model = AutoModelForSequenceClassification.from_pretrained("edumunozsala/roberta_bne_sentiment_analysis_es")

# Texto de ejemplo
text = ("Incumplimiento de fecha de entrega del producto solicitado con casi 2 meses de antelación. "
        "Fue solicitado en Black Friday con precio rebajado y ahora dicen que no hay y llevan con mi dinero casi 2 meses. "
        "Perdí la oportunidad de comprarlo en otro comercio y tengo que decir, que el producto que pedí si hay, "
        "pero ya subido de precio y ahora a Amazon no le viene bien, claro.")

# Preparar el texto para el modelo
input_ids = torch.tensor(tokenizer.encode(text)).unsqueeze(0)

# Obtener los logits del modelo
outputs = model(input_ids)
logits = outputs.logits

# Determinar la clase con la puntuación más alta
predicted_class_idx = logits.argmax(1).item()

# Mapear el índice de clase a una etiqueta legible
labels = ['negativo', 'neutral', 'positivo']
predicted_label = labels[predicted_class_idx]

print(f"El sentimiento de la oración es: {predicted_label}")
```
